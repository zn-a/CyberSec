## Part 1: Network traffic time-series analysis




Load CSV file and select the protocols


```python
file_loc = 'dataset/goose.csv'
csv_data = pd.read_csv(file_loc)

# Convert 'Time' column to datetime
csv_data['Time'] = pd.to_datetime(csv_data['Time'])

# Calculate time difference from the first row in milliseconds
time_diff_ms = (csv_data['Time'] - csv_data['Time'].iloc[0]).dt.total_seconds() * 1000

# Replace 'Time' column with time difference in milliseconds
csv_data['Time'] = time_diff_ms

# Select rows where 'Protocol' is 'GOOSE'
goose = csv_data[csv_data['Protocol'].str.contains('GOOSE')]
```

```python
# Plot the data for all traffic
plt.scatter(csv_data['Time'], csv_data['Length'], s=10)
plt.xlabel('Time (milliseconds)')
plt.ylabel('Length')
plt.title('Time Series: Length of All Traffic')
plt.grid(True)
plt.show()
```

Visualize clusters of the protocols by packet length and and arrival times

```python
# Grouping the data by 'Protocol' and calculating the median for 'Time' and 'Length'
median_data = csv_data.groupby('Protocol')[['Time', 'Length']].median().reset_index()
print(median_data)
# Plot the clusters
sns.scatterplot(data=median_data, x='Time', y='Length', hue='Protocol', style='Protocol', s=100)
plt.title('Clusters of Protocols with Median Time vs Length')
plt.xlabel('Median Time [ms]')
plt.ylabel('Median Length [bytes]')
# plt.xscale('log')
# plt.yscale('log')
# Adjusting legend to fit the plot
plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1), borderaxespad=0)
plt.show()
```

```python
# Filter the DataFrame to include only GOOSE, HTTP, TCP, and UDP protocols
selected_protocols = ['GOOSE', 'HTTP', 'TCP', 'UDP']
filtered_data = csv_data[csv_data['Protocol'].isin(selected_protocols)]

# Iterate over selected protocols and plot CDF for each
for protocol in selected_protocols:
    data = filtered_data[filtered_data['Protocol'] == protocol]['Time']
    sns.ecdfplot(data=data, linestyle='-', linewidth=2, label=protocol)
plt.title('CDF of Arrival Times for Selected Protocols')
plt.xlabel('Time [ms]')
plt.ylabel('CDF')
plt.xscale('log')
# Show legend
plt.legend(title='Protocol', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.show()
```

### Resample Time Series Data

```python
# Select two column only Time and Length
s_csv_data = csv_data[['Time', 'Length']]
s_goose = goose[['Time', 'Length']]

# Set time unit to ms
s_csv_data['Time'] = pd.to_datetime(s_csv_data['Time'], unit='ms')
s_goose['Time'] = pd.to_datetime(s_goose['Time'], unit='ms')
# Set 'Time' column as the index
s_csv_data.set_index('Time', inplace=True)
s_goose.set_index('Time', inplace=True)

# Resample the DataFrame into 100 ms intervals and sum the 'Length' values
s_csv_data = s_csv_data.resample('100L').sum()
s_goose = s_goose.resample('100L').sum()

# Fill NaN values with 0 (in case there are no data points in some intervals)
s_csv_data['Length'] = s_csv_data['Length'].fillna(0)
s_goose['Length'] = s_goose['Length'].fillna(0)

# Reset index to turn the datetime index into a column
s_csv_data.reset_index(inplace=True)
s_goose.reset_index(inplace=True)

# Calculate time difference from the first row in milliseconds
s_csv_data['Time'] = (s_csv_data['Time'] - s_csv_data['Time'].iloc[0]).dt.total_seconds() * 1000
s_goose['Time'] = (s_goose['Time'] - s_goose['Time'].iloc[0]).dt.total_seconds() * 1000
```

```python
# Ensure time is datetime
goose["Time"] = pd.to_datetime(goose["Time"], errors="coerce")
normal["Time"] = pd.to_datetime(normal["Time"], errors="coerce")

# Set as index
goose = goose.set_index("Time")
normal = normal.set_index("Time")

os.makedirs("plots/part-1/resampling", exist_ok=True)

# Define resampling intervals in milliseconds
resample_ms = [100, 200, 500, 1000, 2000]

# Resample and plot
for ms in resample_ms:
    interval = f"{ms}ms"

    goose_resampled = goose.resample(interval).sum(numeric_only=True)
    normal_resampled = normal.resample(interval).sum(numeric_only=True)

    fig, ax = plt.subplots(figsize=(10, 5))
    goose_resampled["Length"].plot(ax=ax, label="GOOSE", alpha=0.75)
    normal_resampled["Length"].plot(ax=ax, label="Normal", alpha=0.75)

    ax.set_title(f"Resampled Packet Lengths @ {interval}")
    ax.set_ylabel("Total Packet Length")
    ax.set_xlabel("Time")
    ax.grid(True, alpha=0.3)
    ax.legend()
    plt.tight_layout()

    plt.savefig(f"plots/part-1/resampling/resample-{ms}ms.svg", format="svg")
    plt.show()
```

```python
# Reload both datasets
goose = pd.read_csv("dataset/goose.csv")
normal = pd.read_csv("dataset/normal.csv")
df_all = pd.concat([goose, normal], ignore_index=True)

df_all["Time"] = pd.to_datetime(df_all["Time"], errors="coerce")
df_all = df_all.dropna(subset=["Time"])
df_all = df_all.set_index("Time")

# Filter valid protocols and resample every 200ms
interval = "200ms"
resampled = df_all.groupby("Protocol").resample(interval).sum(numeric_only=True).reset_index()

# Plot
plt.figure(figsize=(12, 6))
for proto in resampled["Protocol"].unique():
    subset = resampled[resampled["Protocol"] == proto]
    plt.plot(subset["Time"], subset["Length"], label=proto, alpha=0.7)

plt.title("Resampled Packet Lengths by Protocol @ 200ms")
plt.xlabel("Time")
plt.ylabel("Total Packet Length")
plt.grid(True, alpha=0.3)
plt.legend(loc="upper right", fontsize="small", ncol=3)
plt.tight_layout()

plt.savefig("plots/part-1/resampling/resample-200ms-all-protocols.svg", format="svg")
plt.show()
```

### Part 2: Traffic Dispersion Graph
For this part, limit the focus to `GOOSE` protocol

1. Plot graphs from the CSV files using `Networkx`. Sources and destinations in the CSV files serve as edge lists for the `Networkx` graph.

2. Modify the graph by adding switch nodes. All the traffic flows via the switch node.

3. How many total nodes/devices exist for each graph?

Briefly identify and explain the nodal characteristics for each graph 🕸

```python
os.makedirs("plots/part-2", exist_ok=True)
df = pd.read_csv("dataset/goose.csv")
df = df[["Source", "Destination", "Protocol"]]

# Keep only GOOSE traffic
df = df[df["Protocol"] == "GOOSE"]

# Remove duplicates
edges = df[["Source", "Destination"]].drop_duplicates()

# Create a directed graph
G = nx.DiGraph()
G.add_edges_from(edges.to_records(index=False))

# Draw the base graph
plt.figure(figsize=(8, 6))
pos = nx.spring_layout(G, seed=42)
nx.draw(G, pos, with_labels=True, node_color="skyblue", edge_color="gray", node_size=1000, arrows=True)
plt.title("GOOSE Traffic Flow Graph")
plt.tight_layout()
plt.savefig("plots/part-2/graph-base.svg", format="svg")
plt.show()

# Optional: print summary
print(f"Total nodes: {G.number_of_nodes()}")
print(f"Total edges: {G.number_of_edges()}")
```

```python
SWITCH_NODE = "SWITCH"

G_switched = nx.DiGraph()

# Add switch edges: source -> SWITCH -> destination
for src, dst in edges.to_records(index=False):
    G_switched.add_edge(src, SWITCH_NODE)
    G_switched.add_edge(SWITCH_NODE, dst)

# Draw the modified graph
plt.figure(figsize=(8, 6))
pos = nx.spring_layout(G_switched, seed=42)
nx.draw(G_switched, pos, with_labels=True, node_color="lightgreen", edge_color="gray", node_size=1000, arrows=True)
plt.title("GOOSE Traffic via Switch Node")
plt.tight_layout()
plt.savefig("plots/part-2/graph-with-switch.svg", format="svg")
plt.show()

print(f"Total nodes (with switch): {G_switched.number_of_nodes()}")
print(f"Total edges (with switch): {G_switched.number_of_edges()}")
```
